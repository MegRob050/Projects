```python
# Locating correlations between pct_raised and other variables

CR=crowd.corr()
CR.corr()

# fb_total_likes, total_pageviews, and unique_visitors have a correlation of over 80% and will be excluded
```

```python
# Regression model for pct_raised
# Project ID is randomly assigned and can be excluded from the regression model
# Success can also be excluded as it is equal to one and its t value is 100, yet it has a p>|t| value of 0.000

Y=crowd2.pct_raised
X=crowd2[['goal','featured','fb_total_likes','all_or_nothing','total_pageviews','unique_visitors','start_month','end_month','prior_campaigns','num_perks']]
model=sm.OLS(Y,X)
result=model.fit()
result.summary()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
variables=result.model.exog
vif=[variance_inflation_factor(variables, i) for i in range(variables.shape[1])]
vif
```

```python
# All variables in the first regression model had a p>|t| value of 0.00
# Total_pageviews was excluded due to a high vif value of 123.425
# This is the final model. The removal of any more variables result in a lower goodness-of-fitness.
## With any additional changes, the model's goodness-of-fitness will drop below 0.807.
## With this model, the variables result in 80.7% of movement in pct_raised.

Y=crowd2.pct_raised
X=crowd2[['goal','featured','fb_total_likes','all_or_nothing','unique_visitors','start_month','end_month','prior_campaigns','num_perks']]
model=sm.OLS(Y,X)
result=model.fit()
result.summary()

### The most impactful coefs of pct_raised are: 
## Goal with a coef of -6.269e-07. A 1 point increase in goal results in a decrease of pct_raised by 0.00006269%.
## Featured with a coef of 0.5695. A 1 point increases in featured results in an increase of pct_raised by 56.95%.
## All_or_nothing with a coef of 0.2979. A 1 point increase in all_or_nothing results in an increase of pct_raised by 29.79%.

### All other variables have similar coefs that increase pct_raised with unique_visitors having the smallest increase in percentage.
```

```python
# New VIF values have no values over 10 so no new variables will be excluded
# As such, the new regression model will be the final model

variables=result.model.exog
vif=[variance_inflation_factor(variables, i) for i in range(variables.shape[1])]
vif
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
```

```python
## Extra Credit
# Tree 1

dt=tree.DecisionTreeClassifier(max_depth=1)
dt=dt.fit(crowd2[['goal','featured','fb_total_likes','all_or_nothing','unique_visitors','start_month','end_month','prior_campaigns','num_perks']],crowd2.success)
fig,ax=plt.subplots(figsize=(10,5))
tree.plot_tree(dt.fit(crowd2[['goal','featured','fb_total_likes','all_or_nothing','unique_visitors','start_month','end_month','prior_campaigns','num_perks']],crowd2.success),feature_names=['goal','featured','fb_total_likes','all_or_nothing','unique_visitors','start_month','end_month','prior_campaigns','num_perks'], class_names=['0','1'],filled=True, rounded=True, ax=ax)
plt.title('Pred 1')
plt.show()
print('Accuracy score is:',dt.score(crowd2[['goal','featured','fb_total_likes','all_or_nothing','unique_visitors','start_month','end_month','prior_campaigns','num_perks']],crowd2.success))
```
